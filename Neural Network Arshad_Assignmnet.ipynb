```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python

# Load the dataset
file_path = r"D:\DS ASSIGNMENTS\Neural networks\Neural networks\Alphabets_data.csv"
data = pd.read_csv(file_path)

# Step 1: Check for missing values
missing_values_summary = data.isnull().sum()

# Handle missing values (if any)
# Since no missing values were found, this step can be skipped.
# However, if missing values exist, you could use the following strategies:
# Example: Fill missing values with the column mean (if necessary)
# data = data.fillna(data.mean())

# Step 2: Separate features and target variable
X = data.drop(columns=['letter'])  # Features
y = data['letter']  # Target

# Step 3: Normalize features
scaler = MinMaxScaler()
X_normalized = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

# Print summaries
print("Missing Values Summary:")
print(missing_values_summary)
print("\nFeature Summary After Normalization:")
print(X_normalized.describe())
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Step 1: Define the ANN model
model = Sequential()

# Input layer and first hidden layer with 64 neurons and ReLU activation
model.add(Dense(units=64, activation='relu', input_shape=(X_normalized.shape[1],)))

# Output layer with the number of classes equal to the unique values in the target variable
# Assuming a classification task, use softmax activation
num_classes = len(y.unique())  # Number of unique classes in the target variable
model.add(Dense(units=num_classes, activation='softmax'))

# Step 2: Compile the model
model.compile(optimizer='adam',  # Adam optimizer
              loss='sparse_categorical_crossentropy',  # Loss function for multi-class classification
              metrics=['accuracy'])  # Metric to track during training

# Step 3: Print model summary
model.summary()

# Step 4: Train the model (replace with actual split data)
# Assuming data is split into X_train, X_test, y_train, y_test
# history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Step 1: Encode the target variable (if not numeric)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)  # Convert target to numeric labels

# Step 2: Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized, y_encoded, test_size=0.2, random_state=42)

# Step 3: Train the ANN model
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))

# Step 4: Make predictions on the test set
y_pred_prob = model.predict(X_test)
y_pred = np.argmax(y_pred_prob, axis=1)  # Convert probabilities to class labels

# Step 5: Evaluate the model
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
import keras_tuner as kt
from tensorflow.keras.optimizers import Adam

# Step 1: Define a function to create the model
def build_model(hp):
    model = Sequential()
    
    # Add input and hidden layers
    model.add(
        Dense(
            units=hp.Int("units_layer_1", min_value=32, max_value=256, step=32),
            activation=hp.Choice("activation_layer_1", ["relu", "tanh"]),
            input_shape=(X_train.shape[1],)
        )
    )
    
    # Additional hidden layers
    for i in range(hp.Int("num_hidden_layers", 1, 3)):
        model.add(
            Dense(
                units=hp.Int(f"units_layer_{i+2}", min_value=32, max_value=256, step=32),
                activation=hp.Choice(f"activation_layer_{i+2}", ["relu", "tanh"])
            )
        )
    
    # Add output layer
    model.add(Dense(num_classes, activation="softmax"))
    
    # Compile the model
    model.compile(
        optimizer=Adam(learning_rate=hp.Choice("learning_rate", [1e-2, 1e-3, 1e-4])),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

# Step 2: Initialize the tuner
tuner = kt.RandomSearch(
    build_model,
    objective="val_accuracy",
    max_trials=10,  # Number of hyperparameter combinations to try
    executions_per_trial=1,
    directory="hyperparameter_tuning",
    project_name="ann_tuning"
)

# Step 3: Run the tuner
tuner.search(X_train, y_train, epochs=10, validation_data=(X_test, y_test), batch_size=32)

# Step 4: Retrieve the best hyperparameters and build the best model
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
best_model = tuner.hypermodel.build(best_hps)

# Train the best model
history = best_model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))

# Evaluate the best model
test_loss, test_accuracy = best_model.evaluate(X_test, y_test)
print(f"Best Model Test Loss: {test_loss:.4f}")
print(f"Best Model Test Accuracy: {test_accuracy:.4f}")
```
